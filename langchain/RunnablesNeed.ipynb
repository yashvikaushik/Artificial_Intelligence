```python
import random
class nakliLLM(Runnable):

    def __init__(self):
        print('LLM created')


    def invoke(self,prompt):
        response_list=[
            'Delhi is the capital of India',
            'IPL is a cricket league',
            'AI stands for artificial intelligence'
        ]

        return {'response':random.choice(response_list)}
        
        

    def predict(self,prompt):
        response_list=[
            'Delhi is the capital of India',
            'IPL is a cricket league',
            'AI stands for artificial intelligence'
        ]

        return {'response':random.choice(response_list)}
        
```

```python
 llm=nakliLLM()
```

```python
class My_prompt_template(Runnable):
    def __init__(self, template, input_variable):
        self.template = template
        self.input_variable = input_variable

    def invoke(self,input_dictionary):
         return self.template.format(**input_dictionary)

    def format(self, input_dictionary):
        return self.template.format(**input_dictionary)
```

```python
prompt=template = My_prompt_template(
    template="write a {length} poem on {topic}",
    input_variable=['topic']
)

template.format({'length':'short','topic': 'india'})
```

```python
llm.predict(prompt)
```

```python
class nakli_LLM_chain():
    def __init__(self,llm,prompt):
        self.llm=llm
        self.prompt=prompt

    def run(self,input_dict):
        final_prompt=self.prompt.format(input_dict)
        result=self.llm.predict(final_prompt)

        return result['response']
```

```python
template = My_prompt_template(
    template="write a {length} poem on {topic}",
    input_variable=['topic']
)
```

```python
llms=nakliLLM()
```

```python
chain=nakli_LLM_chain(llms,template)
```

```python
chain.run({'length':'short','topic':'india'})
```

```python
from abc import ABC ,abstractclassmethod
```

```python
class Runnable(ABC):

    @abstractclassmethod
    def invoke(input_data):
        pass
        
```

```python
class RunnableConnectorV2(Runnable):
    def __init__(self, runnable_list):
        self.runnable_list = runnable_list

    def invoke(self, input_data):
        for x in self.runnable_list:
            input_data = x.invoke(input_data)
        return input_data
```
